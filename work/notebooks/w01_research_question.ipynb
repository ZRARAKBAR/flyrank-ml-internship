## My Lane (or Freestyle) and Why

I chose the **Content Refresh Prioritization** lane because websites often have many pages, making it difficult to decide which ones should be updated first. My goal is to use machine learning to identify pages that are most likely declining in performance so that content teams can prioritize updates effectively. This project supports data-driven decision making instead of relying only on manual judgment.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## The Question

**Research Question:**
Can machine learning identify webpages that should be refreshed to improve search performance?

**Decision:**
Determine which webpages should be refreshed first.

**Action:**
The content or SEO team will update the highest-priority pages identified by the model.

**Cost of a Wrong Recommendation:**
If declining pages are missed, search traffic may continue to decrease. If healthy pages are selected unnecessarily, time and resources may be wasted.

**Why Machine Learning?**
The dataset contains multiple features such as impressions, CTR, average position, content age, and update history. Machine learning can combine these signals more effectively than simple manual rules.
