## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I selected the Content Refresh Prioritization lane.

This is a classification problem because the goal is to classify webpages as either declining or not declining. The model helps identify which pages should be refreshed first so that the content team can prioritize updates effectively.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


The target is whether a webpage is declining.

The starter dataset uses the **trend_direction** column as the proxy target. Pages where **trend_direction = "down"** are considered declining pages.

This label is based on the observed performance trend in the dataset rather than being manually assigned.

## 3. Success metric

*One metric you can defend. What number means 'good'?*


The success metric is **Precision@50**.

Precision@50 measures how many of the top 50 webpages recommended by the model are actually declining.

A higher Precision@50 means the recommendations are more useful for the content team because they can focus on the pages most likely to need refreshing.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule usually relies on one or two conditions, such as content age or impressions.

Machine learning can analyze multiple features at the same time, including impressions, CTR, average position, content age, and update history.

These factors interact with each other, making the decision more complex than a simple if-statement. Machine learning can identify patterns that improve refresh prioritization and support better decision-making.
